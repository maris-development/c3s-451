The values below define the **WGS84** geographic coordinate system. These values are physical constants (the Earth's radius) used to align your climate data with geographical maps. For standard analysis, leave this cell as is.

```python
# Coordinate system accepted by iris library. Should be similar to the EPSG:4326 systems used for the shapefiles
wgs84 = cs.GeogCS(6378137.0, 6356752.3142)
```

## 3.11 Assess composites of analogues for past and present. 
### This requires initial input to select the analogue variable (from Step 3.10b) and analogue domain (from Step 3.9c, choose the domain corresponding to the analogue variable).  
- a. To produce composite figures (Fig. Analogues.S4 and Fig. 3.2): 
    - i. Uses 1% of analogues 
    - ii. Use time slices of 30 years for ‘past’ and ‘present’: Y1-Y2 is 1950-1980 for ‘past’ and Y3-Y4 is 1995-2025 for ‘present. The ‘present’ time slice potentially needs adjustment (after 2025) 
    - iii. Calculate analogue composites 
- b. Assess analogue composites Z500 or SLP to manually ensure that the analogue composites show analogues which closely resemble the observed event.
    - i. Do subplots b and c look similar to a? Do they show the same large-scale circulation patterns?
    - ii. If not, re-check input variables, and Steps 3.9 and 3.10,
    - iii. Else consider if the event is too dynamically unusual to use analogues.
    - iv. In case the circulation situation is found to be too unique for analogues to describe past and present composites, stop here and describe in the scientific report Section 3.2 that analogues could not be used since the circulation is too unique.
- c. Plot Z500, SLP, tp, and either t2m (for precipitation and Tmean analyses) or Tmax or Tmin of event, past composite, present day composite and difference including significance.
    - i. Check the final text output to ensure correct inputs
    - ii. Interpretate results:
        - 1. Focus on the chosen circulation variable and the event variable, the other figures may give some guidance but do not need interpretation

## 3.12 Assess change in frequency. 
### This requires input of initial variables (date, analogue domain, circulation variable).  
- a. Produce figure of change in frequency of similar circulation analogues (Fig. 3.3; See Figures description section) 
    - i. Calculate Euclidean distance to all events and identify value of closest analogues within all data for 5%, 10% and 20% thresholds.
    - ii. Produce figure of frequencies of closest analogues of 5%, 10%, 20%, including 10-year rolling mean, linear trend and p-values. Interpretation of the figure: 
        - 1. Trend in similar circulation days per year. E.g., trend = 0.07: The frequency of the top 20% most similar circulation events is increasing by about 0.07 days per year, or 0.7 days per 10 years  
        - 2. p < 0.05 means statistically significant, p > 0.05 means not statistically significant. E.g., p=0.11: no statistically significant change in the frequency of the top 20% most similar circulation event 
- b. Copy this Figure to the scientific report to become the second analogues figure 
- c. Note the final text output, providing guidance for the report
    - i. Increasing / No trend / Decreasing
    - ii. Significant / Not significant 
- d. Edit report text to detail the conclusions (see Report template)

### <font color="orange"> User Input </font>

```python
# Start and end of period for frequency assessed (as opposed to periods in Step 3.11): adjust end year
Y1 = 1950 # timeslice 1 start year
Y2 = 2025 # timeslice 1 end year
N = Analogues.number_of_analogues(Y1, Y2, months)
```

### <font color="green"> Autorun Cells </font>

```python
# Find event data
event_cube = Analogues.extract_region(Analogues.extract_date_v2(cube_map_daily[ana_var], event_date), R1)

# Find euclidean distance to all events
event_cube = event_cube - event_cube.collapsed(['latitude', 'longitude'], iris.analysis.MEAN)

P = Analogues.extract_region(cube_map_daily[ana_var], R1)

P = P - P.collapsed(['latitude', 'longitude'], iris.analysis.MEAN)
```

```python
D = []
for Y in np.arange(Y1, Y2):
    D.append(Analogues.euclidean_distance(P.extract(iris.Constraint(year = Y)), event_cube))
x = []
for v in np.arange(len(D)):
    x.append(np.max(D[v]))
ED_max = np.max(x)
S = []
for v in np.arange(len(D)):
    S.append(1-D[v]/np.max(D[v]))
    
# Thresholds - identify value of different percentiles within all data
Sout = np.concatenate(S).ravel()
per5 = np.percentile(Sout,95) # upper 5%
per10 = np.percentile(Sout,90) # upper 10%
per20 = np.percentile(Sout,80) # upper 20%
```

```python
# Thresholds - count for each year
yr5 = []; yr10 = []; yr20 = []
for each in np.arange(len(S)):
    j = S[each][:]
    yr5.append(sum(i > per5 for i in j))
    yr10.append(sum(i > per10 for i in j))
    yr20.append(sum(i > per20 for i in j))

roll5 = []; roll10 = []; roll20 = []
for i in np.arange(len(yr5)-10):
    roll5.append(np.mean(yr5[i:i+10]))
    roll10.append(np.mean(yr10[i:i+10]))
    roll20.append(np.mean(yr20[i:i+10]))

yr_vals = [yr5, yr10, yr20]
roll_vals = [roll5, roll10, roll20]
```

```python
fig, ax, slope_vals, p_vals = Analogues.plot_frequency_timeseries(yr_vals, roll_vals, Y1, Y2)
slope5, slope10, slope20 = slope_vals
pval5, pval10, pval20 = p_vals
```

### <font color="green"> Summary </font>

```python
print('Summary: Change in frequency of similar circulation days')
print('Event definition:')
print('     Date: ', event_date)
print('Analogue calculation:')
print('     Months assessed: ', months)
print('     Number of analogues: ', N)
print('     Analogue domain: ', R1)
print('     Circulation variable: ', ana_var)
print('Timeframe:')
print('      Past: ',Y1,'-',Y2)

print('Results:')
text20='Upper 5%. trend: '+"%.2f" % round(slope5, 2)+', pval: '+"%.2f" % round(pval5, 2)
print(text20)
text20='Upper 10%. trend: '+"%.2f" % round(slope10, 2)+', pval: '+"%.2f" % round(pval10, 2)
print(text20)
text20='Upper 20%. trend: '+"%.2f" % round(slope20, 2)+', pval: '+"%.2f" % round(pval20, 2)
print(text20)

print('Conclusions:')
if slope5>0 and slope10>0 and slope20>0:
    print('1. Increasing')
elif slope5<0 and slope10<0 and slope20<0:
    print('1. Decreasing')
else:
    print('1. No trend')

n=0
for i in [pval5, pval10, pval20]:
    if i < 0.05:
        n+=1
if n>1:
    print('2. Significant')
else:
    print('2. Not significant')
```
